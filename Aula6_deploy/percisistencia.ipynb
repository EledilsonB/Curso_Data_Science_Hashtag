## Testando a Persistência do Modelo

```python
import pandas as pd
import openpyxl

from sklearn.datasets import fetch_california_housing 

from joblib import load
```

```python
# importando arquivos de teste.
housing = fetch_california_housing(as_frame=True)
print(housing.data.shape, housing.target.shape)
print(housing.feature_names[:])
df = pd.DataFrame(housing.data, columns=housing.feature_names)
df['MedHouseVal'] = housing.target
df.head()
```

```python
# Carregando modelo

reg = load('../modelos/regressao.joblib')
```

```python
display(reg.intercept_)
display(reg.coef_)
```

```python
# Fazendo previsão 

X_teste = df.drop('MedHouseVal', axis=1)

reg.predict(X_teste)
```

```python
# Exportando 

previsao = reg.predict(X_teste)

X_teste['Previsao'] = previsao
X_teste.to_excel('../modelos/previsoes_modelo_regressao.xlsx')
```
