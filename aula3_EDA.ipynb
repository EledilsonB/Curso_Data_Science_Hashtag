# Análise Exploratória de Dados

```python
import pandas as pd
from pandas_profiling import ProfileReport

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../Conceitos_Pandas/train.csv')
df.head()
```

```python
display(df.shape)
display('---'*df.shape[1])
display(df.info())
display('---'*df.shape[1])
display('--- Verificando Dados Nulos ---')
display(df.isnull().sum())
display('---'*df.shape[1])
display('--- Dados Estatísticos ---')
display(df.describe())
display('---'*df.shape[1])
display('--- Verificação de Valores Únicos ---')
display(df.nunique())
```

```python
# Verificando o Histograma das tarifas

fig, ax = plt.subplots()
x = df['Fare']

ax.hist(x, bins=20)
plt.title('Tarifas')
plt.tight_layout()
plt.show()
```

```python
# Filtrando tarifas menores que 100

fig, axes = plt.subplots()
x = df[df.Fare < 100].Fare

plt.hist(x, bins=20, linewidth=2, edgecolor='white')

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots()

box_plot = plt.boxplot(x)

plt.show()
```
