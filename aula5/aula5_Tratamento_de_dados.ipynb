# Tratamento de Dados

```python
import pandas as pd

# Bibliotecas de visualização
import matplotlib.pyplot as plt


# Importando modelos do SciKit-Learn
from sklearn.model_selection import train_test_split

from sklearn.neighbors import KNeighborsClassifier 
from sklearn import tree
from sklearn.linear_model import LogisticRegression

# Importando Matrix de Confusão.
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import confusion_matrix

# Importando acuracy score
from sklearn.metrics import accuracy_score

# Importando score de precisão
from sklearn.metrics import precision_score

# Importando o Recall Score
from sklearn.metrics import recall_score

```

```python
df = pd.read_csv('../train.csv')
df
```

```python
print('qtd AGE:', df.Age.count())
print(891-714)
```

```python


def tratamentoDaBase(base):
    # base = base.drop('Cabin', axis=1)
    colunas = base.dtypes[base.dtypes.values == 'str'].index
    base = base.drop(colunas, axis=1)
    base = base.dropna()

    return base

df = tratamentoDaBase(df)
df.count() # Teste com 714 registros
```

# Avaliando o modelo

```python
# Separando dados de treino e teste
X = df.drop('Survived', axis=1)
y = df.Survived

X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=.2, random_state=42)
```

```python
def classificacao_modelos(X_treino, X_teste, y_treino, y_teste):
    neigh = KNeighborsClassifier(n_neighbors=3).fit(X_treino, y_treino)
    dtree = tree.DecisionTreeClassifier(random_state=0).fit(X_treino, y_treino)
    logistic = LogisticRegression(random_state=0, max_iter=1000).fit(X_treino, y_treino)

    pred_knn = neigh.predict(X_teste)
    pred_tree = dtree.predict(X_teste)
    pred_log = logistic.predict(X_teste)

    # Acurácia
    print('Acurácia dos modelos')
    print('KNN:', accuracy_score(y_teste, pred_knn))
    print('Árvore de Decisão', accuracy_score(y_teste, pred_tree))
    print('Regressão Logistica', accuracy_score(y_teste, pred_log))
    print('\n')

    # Precisão
    print('Precisão dos Modelos')
    print('KNN: ', precision_score(y_teste, pred_knn))
    print('Árvore de Decisão: ', precision_score(y_teste, pred_tree))
    print('Regressão Logística: ', precision_score(y_teste, pred_log))
    print('\n')
    
    # Recall
    print('Recall dos Modelos')
    print('KNN: ', recall_score(y_teste, pred_knn))
    print('Árvore de Decisão: ', recall_score(y_teste, pred_tree))
    print('Regressão Logística: ', recall_score(y_teste, pred_log))
    print('\n')
```

```python
classificacao_modelos(X_treino, X_teste, y_treino, y_teste)
```
